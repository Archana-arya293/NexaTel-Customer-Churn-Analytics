```python
#•	Check for outliers in data_gb_used and negative values in total_amount_inr. Flag them.

# finding outliers using IQR Method

# first locate the column

target_column = ["data_gb_used","total_amount_inr"]
for table_name,df in all_dfs.items():
    for column_name in target_column:
        if column_name in df.columns:
            print(column_name,"found in table:",table_name)

```

```python
# calculating IQR for usage_data

data_usage = all_dfs["usage_data"]["data_gb_used"]

Q1 = data_usage.quantile(0.25)
Q3 = data_usage.quantile(0.75)

IQR = Q3-Q1

print("Q1:",Q1)
print("Q3:",Q3)
print("IQR:",IQR)
```

```python
# calculating lower_bound & upper_bound

lower_bound = Q1-(1.5*IQR)
upper_bound = Q3+(1.5*IQR)

print("lower_bound:",round(lower_bound,2))
print("upper_bound:",round(upper_bound,2))
```

```python
# flag data_usage

data_gb_outlier_flag = ((data_usage<lower_bound) | (data_usage>upper_bound))

all_dfs["usage_data"]["data_gb_outlier_flag"] = data_gb_outlier_flag
all_dfs["usage_data"]["data_gb_outlier_flag"].value_counts()
```

```python
# flag outlier in billing,so we have to check the negative values so we will be using 
#explicit business rules here

negative_amount_flag = (all_dfs["billing"]["total_amount_inr"]<0)

all_dfs["billing"]["negative_amount_flag"] = negative_amount_flag
all_dfs["billing"]["negative_amount_flag"].value_counts()
```

```python
# final summary for both check

usage_outlier_count = data_gb_outlier_flag.sum()
usage_outlier_percentage = round(
    usage_outlier_count / len(all_dfs["usage_data"]) * 100,2)

negative_amount_count = negative_amount_flag.sum()
negative_amount_percentage = round(
    negative_amount_count / len(all_dfs["billing"]) * 100,2)

Outlier_summary = pd.DataFrame({
    "table_name": ["usage_data", "billing"],
    "column_name": ["data_gb_used", "total_amount_inr"],
    "check_rule": ["Outside IQR limits","Value less than 0"],
    "flagged_count": [usage_outlier_count,negative_amount_count],
    "flagged_percentage": [usage_outlier_percentage,negative_amount_percentage]})

Outlier_summary
```

# NexaTel Customer Churn – Phase 1 Data Quality Report

## Executive Summary

The data-quality assessment covered 24 source tables and validated 20 issues recorded in `data_quality_issue_log.csv`. Out of 20 checks, 14 matched the expected results and 6 showed differences, giving an exact-match rate of 70%. All six differences were found in the customers table.

## Key Findings

- Missing values were identified in important customer fields, including 745 missing emails, 380 missing addresses and 954 missing annual-income values.
- Duplicate records were found in customers (76), billing (344) and payments (276).
- Customer foreign keys were valid across subscriptions, billing and usage data. However, 184 payment records contained invoice IDs that were not available in billing.
- Billing validation identified 1,149 unique invoices with incorrect GST, 229 negative invoices and 229 future-dated invoices.
- Subscriptions contained 41 missing plan IDs and 102 invalid subscription-status values.
- A total of 171 extreme usage records exceeded the inferred threshold of 350 GB.

## Differences From the Issue Log

| Issue | Expected | Actual | Difference |
|---|---:|---:|---:|
| Missing email | 760 | 745 | -15 |
| Missing income | 950 | 954 | +4 |
| Invalid email | 285 | 286 | +1 |
| Incorrect PIN | 190 | 191 | +1 |
| Inconsistent capitalisation | 570 | 571 | +1 |
| Mixed date format | 380 | 383 | +3 |

## Recommendations

1. Update the six counts in the issue log and clearly document each validation rule.
2. Remove or investigate duplicate customer, billing and payment records.
3. Standardise email, PIN, city-name and date formats before analysis.
4. Automate checks for primary keys, foreign keys, GST calculations, allowed categories and future dates.
5. Confirm the 350 GB usage threshold with the business team before treating these records as outliers.

## Assumptions

GST and future-date issues were counted at the unique-invoice level. Future dates were validated using 31 March 2026 as the project cutoff. The 350 GB usage threshold was inferred from the data and requires business confirmation.

```python
#Clean and export retention campaigns

retention_clean = all_dfs["retention_campaigns"].copy()

# Clean text columns
retention_text_columns = [
    "campaign_id", "customer_id", "campaign_name",
    "offer_type", "channel", "response"]

for column in retention_text_columns:
    retention_clean[column] = (
        retention_clean[column]
        .astype("string")
        .str.strip())

# Standardize category formatting
for column in ["offer_type", "channel", "response"]:
    retention_clean[column] = (
        retention_clean[column].str.title()
    )

# Convert contact date
retention_clean["contact_date"] = pd.to_datetime(
    retention_clean["contact_date"],
    dayfirst=True,
    errors="coerce")

# Convert retained flag to True/False
retained_values = (
    retention_clean["retained_flag"]
    .astype("string")
    .str.strip()
    .str.lower())

retention_clean["retained_flag"] = retained_values.map({
    "yes": True,
    "no": False,
    "true": True,
    "false": False,
    "1": True,
    "0": False,
    "y": True,
    "n": False}).astype("boolean")

# Remove only completely identical rows
exact_duplicates = retention_clean.duplicated().sum()
retention_clean = retention_clean.drop_duplicates()

# Export
retention_clean.to_csv(
    clean_folder / "retention_campaigns_clean.csv",
    index=False)

print("Clean campaign rows:", len(retention_clean))
print("Exact duplicate rows removed:", exact_duplicates)
print(
    "Missing/unrecognized retained flags:",
    retention_clean["retained_flag"].isna().sum())
print("Unparsed contact dates:", retention_clean["contact_date"].isna().sum())
print("File saved: retention_campaigns_clean.csv")
```
